## Fill NaNs with Zeroes

```python
df.fillna(0, inplace=True)
df.head()
```

## Check if columns are integers

```python
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
# print(numeric_cols)

int_inds = []

for col in numeric_cols:
    print(str(col) + ": " + str((df[col] % 1 == 0).all()))
    int_inds.append((df[col] % 1 == 0).all())

int_cols = [col for col, is_int in zip(numeric_cols, int_inds) if is_int]
```

```python
# Convert appropriate columns to int64
for col in int_cols:
    df[col] = df[col].astype('int64')
```

## Verify values are positive

```python
positive_cols = numeric_cols.copy()
positive_cols.remove('opp_offense_dvoa')
positive_cols.remove('opp_defense_dvoa')
positive_cols.remove('stats_passing_pass_yards')
positive_cols.remove('stats_rushing_rush_yards')
positive_cols.remove('stats_receiving_rec_yards')
positive_cols.remove('fantasy_points_ppr')


for col in positive_cols:
    print(str(col) + ": " + str((df[col] >= 0).all()))
```

# Verify IDs are consistent

```python
def check_consistent_columns(df, col1, col2):
    if (df[col1].isnull() != df[col2].isnull()).any():
        return f"{col1} and {col2} have NOT nulls in the same rows."

    # Get unique pairs of values from the two columns, ignoring nulls
    unique_pairs = df[[col1, col2]].dropna().drop_duplicates()

    # Check if we have an equal number of unique pairs and unique values in col1
    # If not, one key in col1 maps to multiple values in col2, which is inconsistent
    return (len(unique_pairs) == unique_pairs[col1].nunique())
```

```python
# Check consistencies between columns
print('game_id and bta_game_id are consistent:', check_consistent_columns(df, 'game_id', 'bta_game_id'))
print('game_id and game_start_time are consistent:', check_consistent_columns(df, 'game_id', 'game_start_time'))
print('player_id and player_name are consistent:', check_consistent_columns(df, 'player_id', 'player_name'))
print('player_id and player_first_name are consistent:', check_consistent_columns(df, 'player_id', 'player_first_name'))
print('player_id and player_last_name are consistent:', check_consistent_columns(df, 'player_id', 'player_last_name'))
print('player_id and player_position are consistent:', check_consistent_columns(df, 'player_id', 'player_position'))
```

```python
agg_df = df.groupby('player_id')['player_position'].nunique()
non_unique_players = agg_df[agg_df > 1].index.tolist()
# print("Players with multiple positions:", non_unique_players)
```

```python
dup_position_df = df[df['player_id'].isin(non_unique_players)][['player_id', 'player_position']].drop_duplicates()
dup_position_df.sort_values('player_id', inplace=True)
print(dup_position_df)
```
