# Data Cleaning and Validation

```python
for col in df.columns:
    print(col + ": " + str(df[col].isnull().sum()))
```

```python
# Check if nulls are in the same rows
print((df['stats_passing_pass_completions'].isnull() == df['stats_passing_pass_int'].isnull()).all())
print((df['stats_passing_pass_completions'].isnull() == df['stats_passing_pass_attempts'].isnull()).all())

print((df['stats_rushing_rush_yards'].isnull() == df['stats_receiving_rec_td'].isnull()).all())
print((df['stats_rushing_rush_yards'].isnull() == df['stats_rushing_rush_attempts'].isnull()).all())
```

```python
# Check positions of nulls
print(df[df['stats_passing_pass_completions'].isnull()]['player_position'].unique())
print(df[df['stats_rushing_rush_yards'].isnull()]['player_position'].unique())

# All of the nulls are linemen or defense, so this is expected
```

## Fill NaNs with Zeroes

```python
df.fillna(0, inplace=True)
df.head()
```

## Check if columns are integers

```python
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
# print(numeric_cols)

int_inds = []

for col in numeric_cols:
    print(str(col) + ": " + str((df[col] % 1 == 0).all()))
    int_inds.append((df[col] % 1 == 0).all())

int_cols = [col for col, is_int in zip(numeric_cols, int_inds) if is_int]
```

```python
# Convert appropriate columns to int64
for col in int_cols:
    df[col] = df[col].astype('int64')
```

## Verify values are positive

```python
positive_cols = numeric_cols.copy()
positive_cols.remove('opp_offense_dvoa')
positive_cols.remove('opp_defense_dvoa')
positive_cols.remove('stats_passing_pass_yards')
positive_cols.remove('stats_rushing_rush_yards')
positive_cols.remove('stats_receiving_rec_yards')
positive_cols.remove('fantasy_points_ppr')


for col in positive_cols:
    print(str(col) + ": " + str((df[col] >= 0).all()))
```

## Convert game start time to time objects

```python
print(df['game_start_time'].dtype)
df['game_start_time'] = pd.to_datetime(df['game_start_time'])
print(df['game_start_time'].dtype)
```
